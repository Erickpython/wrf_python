```python
#  this function converts WRF (UTC) time to EAT
def format_eat_time(wrf_time_raw):
    ts = pd.to_datetime(str(wrf_time_raw))
    eat_time = ts + timedelta(hours=3)
    return eat_time.strftime('%d %b %Y | %H:%M EAT') 
```
