# SKEW-T USING METPY

# WE WILL NEED 

    P - perturbation pressure
    PB - base pressure
    T - perterbation potentialtemperature 
    QVAPOR - water vapor mixing ratio

```python
# compute pressure == 
pressure = (ds.P  + ds.PB) /100  # HPA
```

```python
print(pressure.dims)
print(pressure)
```

```python
# compute temperature 
theta = ds.T + 300 # K

# convert to actual temp 
temperature = theta * (pressure / 1000) ** 0.286
```

```python
print(temperature.dims)
print(temperature)
```

```python
# a single grid point --FIRST TIME STAMP 

i, j = 50, 50 

p = pressure.isel(Time=0, south_north=i, west_east=j)
t = temperature.isel(Time=0, south_north=i, west_east=j)
qv = ds.QVAPOR.isel(Time=0, south_north=i, west_east=j)
```

```python
# CONVERT TO DEW POINT 

from metpy.calc import dewpoint_from_specific_humidity
from metpy.units import units

td = dewpoint_from_specific_humidity(
    p.values * units.hPa,
    t.values * units.K,
    qv.values 
)
```

```python
# plot the skew-t (metpy)

from metpy.plots import SkewT


fig = plt.figure(figsize=(9, 9))
skew = SkewT(fig)

skew.plot(p, t - 273.15, "r")  # Convert temperature to °C for plotting
skew.plot(p, td, "g")  # Plot dew point in green

skew.ax.set_ylim(1000, 100)  # Set y-axis limits for pressure
skew.ax.set_xlim(-40, 40)  # Set x-axis limits for temperature

skew.plot_dry_adiabats()
skew.plot_moist_adiabats()
skew.plot_mixing_lines()

plt.title("Skew-T Log-P Diagram", fontsize=15)
plt.show()

```

# DRAWING ALL THE TIME STEPS --- LATER COMBINING THEM TO AN ANIMATION 

```python
import matplotlib.pyplot as plt
import cartopy.crs as ccrs
import cartopy.feature as cfeature
from cartopy.mpl.gridliner import LONGITUDE_FORMATTER, LATITUDE_FORMATTER
```

```python
# 1. Setup the figure and axes grid
# 5 rows, 5 columns for your 25 timesteps
fig, axes = plt.subplots(
    nrows=5, ncols=5, 
    figsize=(22, 20), 
    subplot_kw={'projection': ccrs.PlateCarree()}
)

# Flatten the axes array to 1D so we can loop through it easily
axes_flat = axes.flatten()

# 2. Loop through all 25 timesteps
for i in range(25):
    ax = axes_flat[i]
    
    # Extract data for the current timestep
    t2_step = t2.isel(Time=i)
    lats_step = lats.isel(Time=i)
    lons_step = lons.isel(Time=i)
    
    # Set the spatial extent
    ax.set_extent([33.5, 42.5, -4.5, 5.5], crs=ccrs.PlateCarree())
    
    # Plot temperature
    # We save the contourf object (im) to use for the colorbar later
    im = ax.contourf(lons_step, lats_step, t2_step, 
                     transform=ccrs.PlateCarree(), cmap="plasma")
    
    # Add map features (scaled down for subplots)
    ax.add_feature(cfeature.BORDERS, edgecolor="black", linewidth=0.5)
    ax.add_feature(cfeature.COASTLINE, edgecolor="black", linewidth=0.5)
    ax.add_feature(cfeature.RIVERS, edgecolor="blue", linewidth=0.5)
    ax.add_feature(cfeature.LAND, facecolor="lightgray")
    ax.add_feature(cfeature.LAKES, facecolor="blue", alpha=0.3)

    # Gridlines: Only show labels on the left of the grid and bottom of the grid
    gl = ax.gridlines(crs=ccrs.PlateCarree(), draw_labels=True,
                      linewidth=0.5, color='gray', alpha=0.5, linestyle='--')
    gl.top_labels = False
    gl.right_labels = False
    
    # Logic to prevent label clutter
    gl.left_labels = True if i % 5 == 0 else False
    gl.bottom_labels = True if i >= 20 else False
    
    gl.xlabel_style = {'size': 8}
    gl.ylabel_style = {'size': 8}

    # Add a small timestamp title to each subplot
    ax.set_title(f"T = {i}", fontsize=10)

# 3. Add a single shared colorbar
# [left, bottom, width, height] in figure coordinates
cbar_ax = fig.add_axes([0.93, 0.15, 0.02, 0.7]) 
fig.colorbar(im, cax=cbar_ax, label="Temperature (°C)")

plt.suptitle("Temperature Evolution: Timesteps 0 to 24", fontsize=20, y=0.95)

# Adjust layout to prevent overlapping
plt.subplots_adjust(wspace=0.1, hspace=0.2)
plt.show()
```
